```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("dados_investimento_e_vagas_ifes.csv")

alunos_por_funcionario = (
    df.groupby('ano')[['num_vagas_estudantes', 'num_vagas_docente', 'num_vagas_tae']]
    .sum()
    .reset_index()
)

alunos_por_funcionario['proporcao_docentes'] = (
    alunos_por_funcionario['num_vagas_estudantes']
    / alunos_por_funcionario['num_vagas_docente']
)

alunos_por_funcionario['proporcao_tae'] = (
    alunos_por_funcionario['num_vagas_estudantes']
    / alunos_por_funcionario['num_vagas_tae']
)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    alunos_por_funcionario['ano'],
    alunos_por_funcionario['proporcao_docentes'],
    label='Estudantes/Docente',
    color='royalblue',
    linewidth=2
)

plt.plot(
    alunos_por_funcionario['ano'],
    alunos_por_funcionario['proporcao_tae'],
    label='Estudantes/TAE',
    color='darkorange',
    linewidth=2
)

plt.xlabel('Ano')
plt.ylabel('Vagas de estudantes por vaga profissional')
plt.title('Evolução das Proporções de Vagas nas IFES')

plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()
```
