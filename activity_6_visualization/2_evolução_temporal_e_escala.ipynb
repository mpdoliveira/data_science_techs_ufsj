```python
import pandas as pd
```

```python
df = pd.read_csv("dados_investimento_e_vagas_ifes.csv")

evolucao_investimento = (
    df.groupby('ano')[['investimento_total_ifes_bilhoes_brl']]
    .sum()
    .reset_index()
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
# Matplotlib e Seaborn
plt.figure(figsize=(12, 5))

sb.lineplot(
    data=evolucao_investimento,
    x='ano',
    y='investimento_total_ifes_bilhoes_brl',
    marker='o',
    color='royalblue',
    linewidth=2
)

plt.title('Evolução do Investimento Total nas IFES')
plt.xlabel('Ano')
plt.ylabel('Investimento (R$ bilhões)')

plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Testando somente com Matplotlib
plt.figure(figsize=(12, 5))

plt.plot(
    evolucao_investimento['ano'],
    evolucao_investimento['investimento_total_ifes_bilhoes_brl'],
    marker='o',
    linewidth=2
)

plt.title('Evolução do Investimento Total nas IFES')
plt.xlabel('Ano')
plt.ylabel('Investimento (R$ bilhões)')
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
evolucao_vagas_estudantes = (
    df.groupby('ano')[['num_vagas_estudantes']]
    .sum()
    .reset_index()
)
```

```python
# Testando somente com Matplotlib
plt.figure(figsize=(12, 5))

plt.plot(
    evolucao_vagas_estudantes['ano'],
    evolucao_vagas_estudantes['num_vagas_estudantes'],
    marker='o',
    linewidth=2   
)

plt.title('Evolução do Número de Vagas de Estudantes nas IFES')
plt.xlabel('Ano')
plt.ylabel('Número de vagas')
plt.grid(axis='y', alpha=0.3)

plt.show()
```
