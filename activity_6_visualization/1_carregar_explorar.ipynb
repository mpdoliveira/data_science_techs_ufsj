# Exploração dos dados

```python
import pandas as pd
```

```python
df = pd.read_csv("dados_investimento_e_vagas_ifes.csv")
```

### Análise de colunas (headers)

```python
print(df.columns, end="\n\n")
```

### Análise de dimensão (min e max) e tipo de dados (head)

```python
ano = df['ano']
print(f"Ano de referência. Min: {ano.min()} Max: {ano.max()}")
print(ano.head(n=20), end="\n\n")
```

```python
investimento_total_bi = df['investimento_total_ifes_bilhoes_brl']
print(f"Investimento Total em IFES (Bilhões). Min: {investimento_total_bi.min()} Max: {investimento_total_bi.max()}")
print(investimento_total_bi.head(n=20))
```

```python
investimento_assistencia_bi = df['investimento_assistencia_estudantil_bilhoes_brl']
print(f"Investimento em Assistência Estudantil (Bilhões). Min: {investimento_assistencia_bi.min()} Max: {investimento_assistencia_bi.max()}")
print(investimento_assistencia_bi.head(n=20))
```

```python
vagas_estudantes = df['num_vagas_estudantes']
print(f"Número de Vagas para Estudantes. Min: {vagas_estudantes.min()} Max: {vagas_estudantes.max()}")
print(vagas_estudantes.head(n=20))
```

```python
vagas_docentes = df['num_vagas_docente']
print(f"Número de Vagas para Docentes. Min: {vagas_docentes.min()} Max: {vagas_docentes.max()}")
print(vagas_docentes.head(n=20))
```

```python
vagas_tae = df['num_vagas_tae']
print(f"Número de Vagas TAE. Min: {vagas_tae.min()} Max: {vagas_tae.max()}")
print(vagas_tae.head(n=20))
```

```python
vagas_terceirizados = df['num_terceirizados']
print(f"Número de Vagas Terceirizados. Min: {vagas_terceirizados.min()} Max: {vagas_terceirizados.max()}")
print(vagas_terceirizados.head(n=20))
```
